## Example 2 — Hugging Face Embeddings and Food Semantic Search

**Goal:** Use the `all-MiniLM-L6-v2` sentence-transformer model to turn sentences into embeddings, compare their similarity, visualize them, and retrieve relevant food/health documents.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import matplotlib.pyplot as plt

model = SentenceTransformer("all-MiniLM-L6-v2")
print("Model loaded successfully!")
```

```python
sentence = "I enjoy learning about healthy nutrition."
embedding = model.encode(sentence)

print("Sentence:", sentence)
print("Embedding shape:", embedding.shape)
print("First 10 values:", embedding[:10])
```

```python
sentences = [
    "I love eating fresh fruits",
    "Vegetables are important for a healthy diet",
    "The football match was exciting",
    "Healthy meals provide essential nutrients",
    "The basketball team won the game"
]

embeddings = model.encode(sentences)
print("Number of sentences:", len(sentences))
print("Embedding matrix shape:", embeddings.shape)
```

```python
similarity_matrix = cosine_similarity(embeddings)
similarity_df = pd.DataFrame(similarity_matrix, index=sentences, columns=sentences)
display(similarity_df.round(2))
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
reduced = pca.fit_transform(embeddings)

plt.figure(figsize=(9, 6))
plt.scatter(reduced[:, 0], reduced[:, 1])
for i, sent in enumerate(sentences):
    plt.annotate(sent, (reduced[i, 0], reduced[i, 1]), fontsize=8)
plt.title("2D Visualization of Food and Sports Sentence Embeddings")
plt.xlabel("PCA Dimension 1")
plt.ylabel("PCA Dimension 2")
plt.grid(True)
plt.show()
```

```python
documents = [
    "Fruits and vegetables provide important vitamins and minerals.",
    "The weather today is sunny and warm.",
    "A balanced diet contains carbohydrates, proteins, fats, vitamins, and minerals.",
    "I enjoy playing football with my friends.",
    "Drinking enough water is important for good health.",
    "Protein helps the body build and repair muscles."
]

doc_embeddings = model.encode(documents)
print("Documents embedded:", len(documents))
```

```python
def semantic_search(query, top_k=2):
    query_embedding = model.encode([query])
    scores = cosine_similarity(query_embedding, doc_embeddings)[0]
    ranked_idx = scores.argsort()[::-1][:top_k]

    print(f"Query: {query}\n")
    for idx in ranked_idx:
        print(f"Score: {scores[idx]:.3f} | {documents[idx]}")

semantic_search("What foods provide vitamins and minerals?")
print("\n" + "-" * 60 + "\n")
semantic_search("How can I build and repair my muscles?")
```
